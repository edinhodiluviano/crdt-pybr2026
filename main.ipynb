```python
from collections import defaultdict
```

```python
class GCounter:
    def __init__(self, replica_id):
        self.replica_id = replica_id
        self.contador = defaultdict(int)

    def valor(self):
        return sum(self.contador.values())

    def incrementar(self, inc):
        self.contador[self.replica_id] += inc

    def merge(self, other):
        for replica_id in other.contador:
            self.contador[replica_id] = max(
                self.contador[replica_id],
                other.contador[replica_id],
            )
```

```python
a = GCounter('id-a')
b = GCounter('id-b')
print(f'{a.valor()=}    {b.valor()=}')
```

```python
a.incrementar(2)
b.incrementar(3)
print(f'{a.valor()=}    {b.valor()=}')
```

```python
a.merge(b)
b.merge(a)
print(f'{a.valor()=}    {b.valor()=}')
```
